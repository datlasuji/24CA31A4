Exercise 10: Fine-tune a BERT-based Model for Named Entity Recognition (NER)

Aim
To use a pretrained BERT model for Named Entity Recognition (NER) to identify entities such as Person, Organization, Location, Date, and Money from text

```python
pip install transformers torch
```

```python

from transformers import pipeline


print("Loading BERT NER model...")

ner = pipeline(
    "ner",
    model="dslim/bert-base-NER",
    aggregation_strategy="simple"
)


text = "Sundar Pichai works at Google in California. He visited India on January 10."


results = ner(text)

print("\nInput Sentence:")
print(text)

print("\nNamed Entities:")
for entity in results:
    print(
        "Entity:", entity["word"],
        "| Type:", entity["entity_group"],
        "| Score:", round(float(entity["score"]), 3)
    )
```
