Exercise 5: Text Classification model Using Naïve Bayes and SVM  

```python

import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score, confusion_matrix


emails = [
    "Win a free prize now",
    "Claim your free gift",
    "Congratulations you won money",
    "Free offer click now",
    "Meeting scheduled for tomorrow",
    "Please attend the class",
    "Project work is completed",
    "Let us meet in the office",
    "Win cash prize today",
    "Please submit your assignment",
    "Get free money now",
    "The meeting starts at ten"
]

email_labels = [
    "Spam", "Spam", "Spam", "Spam",
    "Ham", "Ham", "Ham", "Ham",
    "Spam", "Ham", "Spam", "Ham"
]

X_train, X_test, y_train, y_test = train_test_split(
    emails, email_labels, test_size=0.25,
    random_state=42, stratify=email_labels
)

tfidf = TfidfVectorizer()
X_train_tfidf = tfidf.fit_transform(X_train)
X_test_tfidf = tfidf.transform(X_test)

nb = MultinomialNB()
nb.fit(X_train_tfidf, y_train)

y_pred = nb.predict(X_test_tfidf)

print("SPAM DETECTION")
print("Accuracy:", accuracy_score(y_test, y_pred))
print("Confusion Matrix:")
print(confusion_matrix(y_test, y_pred, labels=["Ham", "Spam"]))

new_emails = [
    "Win free money now",
    "Please attend the meeting"
]

print("New Email Predictions:")
print(nb.predict(tfidf.transform(new_emails)))


# Activity 2: Sentiment Analysis using SVM

reviews = [
    "The product is excellent",
    "I love this product",
    "Amazing quality and service",
    "Very happy with the purchase",
    "The product is wonderful",
    "This is a great experience",
    "The product is terrible",
    "I hate this product",
    "Very poor quality",
    "I am unhappy with the service",
    "The product is disappointing",
    "This is a bad experience"
]

sentiments = [
    "Positive", "Positive", "Positive",
    "Positive", "Positive", "Positive",
    "Negative", "Negative", "Negative",
    "Negative", "Negative", "Negative"
]

X_train, X_test, y_train, y_test = train_test_split(
    reviews, sentiments, test_size=0.25,
    random_state=42, stratify=sentiments
)

tfidf = TfidfVectorizer()
X_train_tfidf = tfidf.fit_transform(X_train)
X_test_tfidf = tfidf.transform(X_test)

svm = SVC(kernel="linear")
svm.fit(X_train_tfidf, y_train)

y_pred = svm.predict(X_test_tfidf)

print("\nSENTIMENT ANALYSIS")
print("Accuracy:", accuracy_score(y_test, y_pred))
print("Confusion Matrix:")
print(confusion_matrix(
    y_test, y_pred, labels=["Negative", "Positive"]
))

new_reviews = [
    "The product is amazing",
    "The quality is very poor"
]

print("New Review Predictions:")
print(svm.predict(tfidf.transform(new_reviews)))
```
