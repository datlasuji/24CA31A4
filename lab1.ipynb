1. Create a Python list containing the given text corpus.

```python
corpus = [
    "<p>Data Science is AMAZING!!! Visit https://datascience.com</p>",
    "Python 3.12 is released in 2025.",
    "NLP is used in AI.",
    "Contact us at info@gmail.com"
]
```

#### 2. Import the required libraries (re and BeautifulSoup).

```python
import re
from bs4 import BeautifulSoup
```

3. Convert all text to lowercase

```python
for text in corpus:
    text = text.lower()
    print(text)
```

```python
lowercase_corpus = [text.lower() for text in corpus]

print(lowercase_corpus)
```

4. Remove HTML tags.

```python
html_removed = [
    BeautifulSoup(text, "html.parser").get_text()
    for text in lowercase_corpus
]

print(html_removed)
```

5. Remove URLs.

```python
url_removed = [
    re.sub(r'https?://\S+|www\.\S+', '', text)
    for text in html_removed
]

print(url_removed)
```

6. Remove email addresses.

```python
email_removed = [
    re.sub(r'\b[\w\.-]+@[\w\.-]+\.\w+\b', '', text)
    for text in url_removed
]

print(email_removed)
```

7. Remove digits.

```python
digit_removed = [
    re.sub(r'\d+', '', text)
    for text in email_removed
]

print(digit_removed)
```

 8. Remove punctuation and special characters.

```python
punct_removed = [
    re.sub(r'\W+', ' ', text)
    for text in digit_removed
]

print(punct_removed)
```

#### 9. Remove extra white spaces.

```python
cleaned_corpus = [
    re.sub(r'\s+', ' ', text).strip()
    for text in punct_removed
]

print(cleaned_corpus)
```

```python
#### 10. Store the cleaned text in a new list named cleaned_corpus.
```

```python
print("The cleaned text has been stored in the list 'cleaned_corpus'.")
print(cleaned_corpus)
```

#### 11. Display the original corpus and the cleaned corpus.

```python
print("Original Corpus:\n")

for sentence in corpus:
    print(sentence)

print("\nCleaned Corpus:\n")

for sentence in cleaned_corpus:
    print(sentence)
```

#### 12. Compare the text before and after cleaning.

```python
print("Comparison Before and After Cleaning\n")

for original, cleaned in zip(corpus, cleaned_corpus):
    print("Original :", original)
    print("Cleaned  :", cleaned)
    print("-" * 60)
```

### Live Scenario

ABC Electronics is a popular e-commerce company that sells smartphones, laptops, and electronic accessories. Every day, customers post reviews about their shopping experience on the company's website.

Before performing **Sentiment Analysis**, the Data Analytics team must clean the customer reviews because they contain HTML tags, URLs, email addresses, numbers, punctuation marks, emojis, and unnecessary white spaces.

As a Data Scientist, your task is to clean the customer reviews using the text preprocessing techniques learned in this experiment and prepare the dataset for further Natural Language Processing (NLP) tasks.

---

Objectives

- Collect five customer reviews.
- Apply text cleaning techniques.
- Compare the original and cleaned reviews.
- Analyze the impact of text cleaning.

---


 Sample Dataset

```python
# Customer Reviews

reviews = [
    "<p>Excellent phone!!! 😍 Battery backup is amazing. Visit https://abcshop.com</p>",
    "Worst laptop ever!! 😡 I paid $999 in 2025. Contact support@abcshop.com",
    "Fast delivery 🚚 and excellent packaging. Highly Recommended!!!",
    "<div>Camera quality is AWESOME 📸. Worth every penny!!!</div>",
    "Average product 😐 Delivery was delayed by 3 days. Visit www.shippingupdates.com"
]

print("Original Reviews\n")

for review in reviews:
    print(review)
```

#### Import Libraries

```python
import re
from bs4 import BeautifulSoup
```

#### Clean the Reviews

```python
cleaned_reviews = []

for review in reviews:

    # Convert to lowercase
    review = review.lower()

    # Remove HTML tags
    review = BeautifulSoup(review, "html.parser").get_text()

    # Remove URLs
    review = re.sub(r'https?://\S+|www\.\S+', '', review)

    # Remove Email IDs
    review = re.sub(r'\b[\w\.-]+@[\w\.-]+\.\w+\b', '', review)

    # Remove digits
    review = re.sub(r'\d+', '', review)

    # Remove punctuation and special characters
    review = re.sub(r'\W+', ' ', review)

    # Remove extra spaces
    review = re.sub(r'\s+', ' ', review).strip()

    cleaned_reviews.append(review)
```

#### Compare Before and After

```python
print("Comparison of Original and Cleaned Reviews\n")

for original, cleaned in zip(reviews, cleaned_reviews):

    print("Original Review")
    print(original)

    print("\nCleaned Review")
    print(cleaned)

    print("-"*80)
```
