Exercise 3: Text Representation Techniques 

```python

import pandas as pd
import matplotlib.pyplot as plt
from sklearn.feature_extraction.text import CountVectorizer
from wordcloud import WordCloud
from collections import Counter

corpus = [
    "I love machine learning",
    "I love Python programming",
    "Machine learning is interesting"
]


words = sorted(set(" ".join(corpus).lower().split()))
data = []

for word in words:
    vector = [1 if word == w else 0 for w in words]
    data.append(vector)

print("One-Hot Encoding:")
print(pd.DataFrame(data, index=words, columns=words))


cv = CountVectorizer()
X = cv.fit_transform(corpus)

print("\nVocabulary:")
print(cv.get_feature_names_out())

print("\nDocument-Term Matrix:")
print(pd.DataFrame(
    X.toarray(),
    columns=cv.get_feature_names_out()
))


text = " ".join(corpus)

wc = WordCloud(
    width=800,
    height=400,
    background_color="white"
).generate(text)

plt.imshow(wc, interpolation="bilinear")
plt.axis("off")
plt.show()

print("\nTop Five Frequent Words:")
print(Counter(text.lower().split()).most_common(5))
```
