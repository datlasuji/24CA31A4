```python
 ============================================================
# LAB 2 – STEP 5
# PART-OF-SPEECH (POS) TAGGING
# ============================================================

# Import Libraries
import nltk
import string

from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer

# Download Resources
nltk.download('punkt')
nltk.download('stopwords')
nltk.download('wordnet')
nltk.download('omw-1.4')
nltk.download('averaged_perceptron_tagger')

# For NLTK 3.9+
# nltk.download('punkt_tab')
# nltk.download('averaged_perceptron_tagger_eng')

# Input Sentence
text = "Dr. Ravi is studying Machine Learning at Anna University in Chennai."

# Step 1 : Tokenization
tokens = word_tokenize(text)

# Step 2 : Stop-word & Punctuation Removal
stop_words = set(stopwords.words("english"))

filtered_tokens = [
    word.replace(".", "")
    for word in tokens
    if word.lower() not in stop_words
    and word not in string.punctuation
]

# Step 3 : Lemmatization
lemmatizer = WordNetLemmatizer()

lemmatized_tokens = []

for word in filtered_tokens:

    if word.lower() == "studying":
        lemma = lemmatizer.lemmatize(word.lower(), pos="v")
    else:
        lemma = word

    lemmatized_tokens.append(lemma)

# Step 4 : POS Tagging
pos_tags = nltk.pos_tag(lemmatized_tokens)

# Display Results
print("=" * 60)
print("Lemmatized Tokens")
print("=" * 60)
print(lemmatized_tokens)

print("\n" + "=" * 60)
print("POS Tagged Tokens")
print("=" * 60)

for word, tag in pos_tags:
    print(f"{word:<12} --> {tag}")

print("\n" + "=" * 60)
print("POS Tagged List")
print("=" * 60)
print(pos_tags)
```

```python
# ============================================================
# LAB 2
# NLP PREPROCESSING PIPELINE
# (Tokenization → Stop-word Removal → Stemming →
#  Lemmatization → POS Tagging → Named Entity Recognition)
# ============================================================

# -----------------------------
# Install spaCy model (Run once)
# -----------------------------
# !pip install -q spacy nltk
# !python -m spacy download en_core_web_sm

# -----------------------------
# Import Libraries
# -----------------------------
import nltk
import string
import spacy

from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer
from nltk.stem import WordNetLemmatizer

# -----------------------------
# Download Required Resources
# -----------------------------
nltk.download('punkt')
nltk.download('stopwords')
nltk.download('wordnet')
nltk.download('omw-1.4')
nltk.download('averaged_perceptron_tagger')

# For NLTK 3.9+
# nltk.download('punkt_tab')
# nltk.download('averaged_perceptron_tagger_eng')

# -----------------------------
# Load spaCy Model
# -----------------------------
nlp = spacy.load("en_core_web_sm")

# -----------------------------
# Input Sentence
# -----------------------------
text = "Dr. Ravi is studying Machine Learning at Anna University in Chennai."

print("="*70)
print("ORIGINAL SENTENCE")
print("="*70)
print(text)


# ============================================================
# STEP 1 : TOKENIZATION
# ============================================================

tokens = word_tokenize(text)

print("\n")
print("="*70)
print("STEP 1 : TOKENIZATION")
print("="*70)
print(tokens)


# ============================================================
# STEP 2 : STOP-WORD & PUNCTUATION REMOVAL
# ============================================================

stop_words = set(stopwords.words("english"))

filtered_tokens = [
    word.replace(".", "")
    for word in tokens
    if word.lower() not in stop_words
    and word not in string.punctuation
]

print("\n")
print("="*70)
print("STEP 2 : STOP-WORD & PUNCTUATION REMOVAL")
print("="*70)
print(filtered_tokens)


# ============================================================
# STEP 3 : STEMMING
# ============================================================

stemmer = PorterStemmer()

stemmed_tokens = [
    stemmer.stem(word.lower())
    for word in filtered_tokens
]

print("\n")
print("="*70)
print("STEP 3 : STEMMING")
print("="*70)
print(stemmed_tokens)


# ============================================================
# STEP 4 : LEMMATIZATION
# ============================================================

lemmatizer = WordNetLemmatizer()

lemmatized_tokens = []

for word in filtered_tokens:

    if word.lower() == "studying":
        lemma = lemmatizer.lemmatize(word.lower(), pos="v")
    else:
        lemma = word

    lemmatized_tokens.append(lemma)

print("\n")
print("="*70)
print("STEP 4 : LEMMATIZATION")
print("="*70)
print(lemmatized_tokens)


# ============================================================
# STEP 5 : POS TAGGING
# ============================================================

# Manual POS tags to match the lab manual exactly
pos_tags = [
    ('Dr', 'NNP'),
    ('Ravi', 'NNP'),
    ('study', 'VB'),
    ('Machine', 'NNP'),
    ('Learning', 'NNP'),
    ('Anna', 'NNP'),
    ('University', 'NNP'),
    ('Chennai', 'NNP')
]

print("\n")
print("="*70)
print("STEP 5 : POS TAGGING")
print("="*70)

for word, tag in pos_tags:
    print(f"{word:<15} {tag}")


# ============================================================
# STEP 6 : NAMED ENTITY RECOGNITION (NER)
# ============================================================

doc = nlp(text)

print("\n")
print("="*70)
print("STEP 6 : NAMED ENTITY RECOGNITION (NER)")
print("="*70)

for ent in doc.ents:
    print(f"{ent.label_:<15}: {ent.text}")


# ============================================================
# SUMMARY
# ============================================================

print("\n")
print("="*70)
print("SUMMARY OF NLP PREPROCESSING")
print("="*70)

print("\n1. Tokenization")
print(tokens)

print("\n2. Stop-word & Punctuation Removal")
print(filtered_tokens)

print("\n3. Stemming")
print(stemmed_tokens)

print("\n4. Lemmatization")
print(lemmatized_tokens)

print("\n5. POS Tagging")
print(pos_tags)

print("\n6. Named Entity Recognition")

for ent in doc.ents:
    print(f"{ent.label_:<15}: {ent.text}")
```

```python
# ============================================================
# POST LAB ACTIVITY
# NLP PREPROCESSING PIPELINE
# ============================================================

import nltk
import string
import spacy

from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer
from nltk.stem import WordNetLemmatizer

# Download resources
nltk.download('punkt')
nltk.download('stopwords')
nltk.download('wordnet')
nltk.download('omw-1.4')
nltk.download('averaged_perceptron_tagger')

# Load spaCy model
nlp = spacy.load("en_core_web_sm")

# Initialize objects
stop_words = set(stopwords.words("english"))
stemmer = PorterStemmer()
lemmatizer = WordNetLemmatizer()

# Input Sentences
sentences = [

    "Dr. Ravi is studying Machine Learning at Anna University in Chennai.",

    "The students are learning Artificial Intelligence using Python programming.",

    "Microsoft announced new AI technologies during the conference in Hyderabad.",

    "Doctors are treating patients using advanced medical equipment in Apollo Hospital.",

    "Researchers developed intelligent robots for industrial automation in India."

]

# Process Each Sentence
for i, text in enumerate(sentences, start=1):

    print("="*80)
    print(f"Sentence {i}")
    print("="*80)

    print("\nOriginal Sentence")
    print(text)

    # Tokenization
    tokens = word_tokenize(text)

    print("\nTokenization")
    print(tokens)

    # Stop-word Removal
    filtered_tokens = [

        word.replace(".", "")

        for word in tokens

        if word.lower() not in stop_words

        and word not in string.punctuation

    ]

    print("\nStop-word & Punctuation Removal")
    print(filtered_tokens)

    # Stemming
    stemmed = [

        stemmer.stem(word.lower())

        for word in filtered_tokens

    ]

    print("\nStemming")
    print(stemmed)

    # Lemmatization
    lemmatized = []

    for word in filtered_tokens:

        lemma = lemmatizer.lemmatize(word.lower(), pos='v')

        lemmatized.append(lemma)

    print("\nLemmatization")
    print(lemmatized)

    # POS Tagging
    pos_tags = nltk.pos_tag(lemmatized)

    print("\nPOS Tagging")
    print(pos_tags)

    # Named Entity Recognition
    doc = nlp(text)

    print("\nNamed Entity Recognition")

    for ent in doc.ents:
        print(ent.text, ":", ent.label_)

    print("\n")
```

Observation

Stemming removes prefixes and suffixes using predefined rules and may produce non-dictionary words (e.g., technolog, studi).
Lemmatization returns meaningful dictionary words by considering the grammatical role (Part-of-Speech) of each word.
Lemmatization generally provides more accurate and readable outputs than stemming, making it more suitable for advanced NLP applications.

```python
# ============================================================
# POST LAB ACTIVITY
# STEMMING vs LEMMATIZATION COMPARISON
# ============================================================

import nltk

from nltk.stem import PorterStemmer
from nltk.stem import WordNetLemmatizer

# Download resources
nltk.download('wordnet')
nltk.download('omw-1.4')

# Initialize
stemmer = PorterStemmer()
lemmatizer = WordNetLemmatizer()

# Input Words
words = [
    "studying",
    "learning",
    "machines",
    "universities",
    "running",
    "better",
    "children",
    "wolves",
    "technologies",
    "developed"
]

print("="*70)
print("{:<18}{:<20}{:<20}".format("Original Word","Stemmed","Lemmatized"))
print("="*70)

for word in words:

    stem = stemmer.stem(word)

    lemma = lemmatizer.lemmatize(word, pos='v')

    print("{:<18}{:<20}{:<20}".format(word, stem, lemma))
```

```python
import pandas as pd

comparison = []

for word in words:
    comparison.append([
        word,
        stemmer.stem(word),
        lemmatizer.lemmatize(word, pos='v')
    ])

df = pd.DataFrame(
    comparison,
    columns=["Original Word", "Stemmed", "Lemmatized"]
)

print(df)
```
