Exercise 8: Train and Evaluate Word2Vec (CBOW) Embeddings for Semantic Similarity Analysis

```python

import numpy as np
from collections import Counter


corpus = [
    "machine learning is useful",
    "machine learning is interesting",
    "artificial intelligence is useful",
    "artificial intelligence uses machine learning",
    "deep learning is useful",
    "machine learning helps computers",
    "artificial intelligence is interesting",
    "deep learning uses machine learning",
    "natural language processing uses machine learning",
    "artificial intelligence and machine learning are useful"
]

tokenized_corpus = [sentence.lower().split() for sentence in corpus]

print("ACTIVITY 1: TOKENIZED CORPUS")
for sentence in tokenized_corpus:
    print(sentence)


vocab = sorted(set(word for sentence in tokenized_corpus
                   for word in sentence))
word_to_id = {word: i for i, word in enumerate(vocab)}
vocab_size = len(vocab)

print("\nACTIVITY 2: CBOW MODEL")
print("Vocabulary:", vocab)


window = 2
pairs = []

for sentence in tokenized_corpus:
    for i, target in enumerate(sentence):
        context = sentence[max(0, i-window):i] + \
                  sentence[i+1:i+window+1]
        for word in context:
            pairs.append((word_to_id[word], word_to_id[target]))


np.random.seed(42)
dimension = 50
W1 = np.random.randn(vocab_size, dimension) * 0.1
W2 = np.random.randn(dimension, vocab_size) * 0.1
learning_rate = 0.03

for epoch in range(100):
    np.random.shuffle(pairs)

    for context_id, target_id in pairs:
        hidden = W1[context_id].copy()
        scores = hidden @ W2
        scores -= np.max(scores)

        probs = np.exp(scores)
        probs /= np.sum(probs)

        probs[target_id] -= 1
        grad_W2 = np.outer(hidden, probs)
        grad_W1 = W2 @ probs

        W2 -= learning_rate * grad_W2
        W1[context_id] -= learning_rate * grad_W1

print("CBOW model trained successfully.")
print("Vocabulary size:", vocab_size)

# Activity 3: Generate Word Embeddings
print("\nACTIVITY 3: WORD EMBEDDINGS")

print("Vector for Machine:")
print(np.round(W1[word_to_id["machine"]], 3))

print("\nVector for Learning:")
print(np.round(W1[word_to_id["learning"]], 3))

print("\nVector Dimension:", dimension)


def similarity(word1, word2):
    v1 = W1[word_to_id[word1]]
    v2 = W1[word_to_id[word2]]
    denominator = np.linalg.norm(v1) * np.linalg.norm(v2)

    if denominator == 0:
        return 0.0

    return float(np.dot(v1, v2) / denominator)

print("\nACTIVITY 4: SEMANTIC SIMILARITY")

learning_vector = W1[word_to_id["learning"]]
scores = []

for word in vocab:
    if word != "learning":
        score = similarity("learning", word)
        scores.append((word, score))

scores.sort(key=lambda x: x[1], reverse=True)

print("Top 5 words similar to Learning:")
for word, score in scores[:5]:
    print(word, round(score, 4))

print("\nMachine-Learning Similarity:",
      round(similarity("machine", "learning"), 4))

print("Artificial-Intelligence Similarity:",
      round(similarity("artificial", "intelligence"), 4))
```
