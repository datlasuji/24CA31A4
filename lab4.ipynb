Exercise 4: TF-IDF (Term Frequency–Inverse Document Frequency)

```python
import pandas as pd
import numpy as np
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer


corpus = [
    "I love machine learning",
    "I love Python programming",
    "Machine learning is interesting"
]


cv = CountVectorizer()
count_matrix = cv.fit_transform(corpus)
words = cv.get_feature_names_out()

print("Vocabulary:")
print(words)


tf = count_matrix.toarray()
total_words = tf.sum(axis=1)

tf_matrix = tf / total_words[:, None]

print("\nTotal words in each document:")
print(total_words)

print("\nTF Matrix:")
print(pd.DataFrame(tf_matrix, columns=words))


df = (tf > 0).sum(axis=0)

N = len(corpus)
idf = np.log(N / df)

print("\nDF and IDF:")
print(pd.DataFrame({
    "Word": words,
    "DF": df,
    "IDF": idf
}))


tfidf = TfidfVectorizer()
tfidf_matrix = tfidf.fit_transform(corpus)

print("\nTF-IDF Matrix:")
print(pd.DataFrame(
    tfidf_matrix.toarray(),
    columns=tfidf.get_feature_names_out()
).round(3))
```
